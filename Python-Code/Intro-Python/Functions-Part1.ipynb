## 1. What is a function?

A **function** is a named piece of code that performs a task.

Instead of writing the same instructions repeatedly, we can put those instructions
in a function and use it whenever we need it.

Python provides many **built-in functions**. You have already used some of them.
Run each example below.


````python
print("Welcome to Python!")
````

````python
type(250)
````

````python
float("250.75")
````

| Function | What it does |
|---|---|
| `print()` | Displays information |
| `type()` | Returns the type of a value |
| `int()` | Converts a suitable value into an integer |
| `float()` | Converts a suitable value into a floating-point number |
| `input()` | Gets text entered by the user |

Using a function is called **calling the function**.

We call a function by writing its name followed by parentheses:

```python
print("Hello!")
```

The value inside the parentheses is an **argument**.
Here, `"Hello!"` is the argument passed to `print()`.

**Notebook note:** Colab displays the value of a final expression such as `type(250)`.
Use `print()` when you want your program to explicitly display information.

## 3. Getting information from the user with `input()`

So far, we have written values directly in our programs:

```python
name = "Alex"
```

The `input()` function lets the person running the program enter a value.


````python
name = input("Enter your name: ")
print("Hello,", name)
````

When you run the cell above:

1. Python displays the message `Enter your name:`.
2. The program waits for you to type a response.
3. You type your response and press **Enter**.
4. Python stores the response in `name`.
5. The next line displays your greeting.

If you enter `Alex`, the greeting is:

```text
Hello, Alex
```

The message inside `input()` is called a **prompt**. It tells the user what to enter.

### Important: `input()` always returns a string

This is true even when the user types a number.
Run the next cell and enter `500`.


````python
deposit = input("Enter your deposit amount: ")

print("You entered:", deposit)
print("Data type:", type(deposit))
````

If you enter `500`, the output is:

```text
You entered: 500
Data type: <class 'str'>
```

Python stores `"500"` as text, not as the number `500`.


## 4. Converting user input into numbers

To use a response in a numeric calculation, we usually need to convert it into a number.

- Use `int()` for whole-number input, such as a number of months.
- Use `float()` for input that may contain decimals, such as a deposit amount.

### Example: A deposit amount

Start with separate steps. Run the next cell and enter `450.50`.


````python
deposit_text = input("Enter your deposit amount: ")
deposit = float(deposit_text)

new_balance = deposit + 100

print("Balance after adding 100:", new_balance)
````

If you enter `450.50`, the output is:

```text
Balance after adding 100: 550.5
```

| Code | Meaning |
|---|---|
| `input(...)` | Gets the user's response as text |
| `float(deposit_text)` | Converts that text into a number |
| `deposit + 100` | Performs numeric addition |

Once you understand the separate steps, you can combine input and conversion:


````python
deposit = float(input("Enter your deposit amount: "))
print("Deposit:", deposit)
````

Python first calls `input()`. It then passes the returned text to `float()`.

### Example: A whole number


````python
months = int(input("Enter the number of months: "))
print("Number of months:", months)
````

For these examples, enter numbers such as `500` or `450.50`.
Do not include dollar signs or commas. For `int()`, enter a whole number such as `12`.

## 5. Defining and calling your own function

Built-in functions are provided by Python. We can also create our own functions
using the keyword **`def`**.

We will learn two common ways to write a function:

1. **A function without an explicit return value:** it performs a task, such as
   displaying a message with `print()`.
2. **A function that returns a value:** it uses `return` to send a result back
   so we can store or use that result.

### Example A: A function without an explicit return value

Run the definition below.

NOTE: **Indentation** is required for the code block used in a function!


````python
def show_welcome():
    print("Welcome to the savings calculator!")
    print("Let us calculate your savings.")
````

The cell above **defines** the function. It does not yet run the instructions
inside it, so no message is displayed just from defining it.

To run those instructions, **call the function**:


````python
show_welcome() #calling the function using the function name!!!
````

Expected output:

```text
Welcome to the savings calculator!
Let us calculate your savings.
```

This function performs a task: displaying messages. It has no `return` statement.
We can call it again whenever we need to display the same messages.

### Understanding the definition

```python
def show_welcome():
    print("Welcome to the savings calculator!")
```

- `def` begins a function definition.
- `show_welcome` is the function's name.
- Parentheses are required. Here, they are empty because no arguments are needed.
- The colon `:` begins the function's code block.
- The indented statements form the **function body**.

Just as with `if` statements, indentation matters. Use four spaces for the function body.

**Remember:** Defining a function makes it available. Calling it runs its body.

### Example B: A function that returns a value

Suppose we want a function to provide a bonus amount that we can use in a calculation.
This function uses `return` to send back the number `100`:


````python
def get_bonus():
    return 100
````

Call the function and store its returned value in a variable:


````python
bonus = get_bonus()
print("Bonus:", bonus)
````

Expected output:

```text
Bonus: 100
```

The value returned by `get_bonus()` is stored in `bonus`.
We can now use that value in another calculation:


````python
balance = 900 + bonus
print("Balance with bonus:", balance)
````

Expected output:

```text
Balance with bonus: 1000
```

`print()` displays information. `return` sends a value back to the calling code.

**Small Python detail:** A function without an explicit return value automatically
returns `None`, which represents the absence of a value. These are two common
ways to use functions; a function can also both display information and return a value.


## 6. Parameters and arguments

A function can receive values and use them to perform its task.


````python
def greet_customer(name): #parameter is name
    print("Hello,", name)
````

Now call it with different values:


````python
greet_customer("Maya") #the function is called with the argument "Maya"
greet_customer("Daniel")
````

Expected output:

```text
Hello, Maya
Hello, Daniel
```

The same function works with different names.

| Term | Meaning | In this example |
|---|---|---|
| **Parameter** | A name in the function definition that receives a value | `name` |
| **Argument** | A value supplied when calling the function | `"Maya"` |

```python
def greet_customer(name):    # name is a parameter
    print("Hello,", name)

greet_customer("Maya")       # "Maya" is an argument
```

A variable can also supply an argument:


````python
customer = "Daniel"
greet_customer(customer)
````

The value stored in `customer` is passed to the function.

**A parameter does not automatically ask the user a question.**
To ask the user for information, we explicitly call `input()`:


````python
customer = input("Enter your name: ")
greet_customer(customer)
````

## 7. Returning a result with `return`

Now combine **parameters** with **`return`** to calculate a result from values
passed into a function.

Consider a function that calculates interest for one year:


````python
def calculate_interest(principal, rate):
    interest = principal * rate
    return interest
````

In this example:

- `principal` is the initial deposit amount.
- `rate` is the annual interest rate written as a decimal.
- `0.04` represents 4%.
- The calculation assumes **one year of simple interest**.

Call the function and store its result:


````python
earned_interest = calculate_interest(1000, 0.04)

print("Interest earned:", earned_interest)
````

Expected output:

```text
Interest earned: 40.0
```

The steps are:

1. `1000` is passed to `principal`.
2. `0.04` is passed to `rate`.
3. The function calculates `1000 * 0.04`.
4. `return` sends `40.0` back.
5. The returned value is stored in `earned_interest`.

Because we have the returned value, we can use it in another calculation:


````python
ending_balance = 1000 + earned_interest
print("Ending balance:", ending_balance)
````

Expected output:

```text
Ending balance: 1040.0
```

**A `return` statement also immediately ends the current function call.**


## 8. Putting input, a function, and output together

We can now build a complete interactive program.
Run the cell below and enter `2000`.


````python
def calculate_interest(principal, rate):
    return principal * rate


deposit = float(input("Enter your deposit amount: "))

interest = calculate_interest(deposit, 0.04)
ending_balance = deposit + interest

print("Interest earned in one year:", interest)
print("Balance after one year:", ending_balance)
````

If the user enters `2000`, the output is:

```text
Interest earned in one year: 80.0
Balance after one year: 2080.0
```

Notice the responsibilities of each function:

- `input()` gets the response.
- `float()` converts it into a number.
- `calculate_interest()` performs our calculation.
- `print()` displays the results.

## 9. More examples with functions

Functions can contain the `if`, `elif`, and `else` statements you already know.

Suppose a **fictional account** has this monthly fee rule:

- Balance of at least `1000`: no fee.
- Balance below `1000`: a fee of `5`.


````python
def calculate_monthly_fee(balance):
    if balance >= 1000:
        return 0
    else:
        return 5
````

Test the function:


````python
print(calculate_monthly_fee(1500))
print(calculate_monthly_fee(600))
print(calculate_monthly_fee(1000))
````

Expected output:

```text
0
5
0
```

Notice the indentation:

- The `if` and `else` statements are inside the function.
- Each `return` statement is inside its corresponding branch.

We can also use the function with user input:


````python
account_balance = float(input("Enter your account balance: "))

fee = calculate_monthly_fee(account_balance)

print("Monthly fee:", fee)
````

## 10. Short extension: Default parameter values

A parameter can have a **default value**.
Python uses that value when the caller does not supply an argument for that parameter.

We can update our interest function to use a default rate of 4%:


````python
def calculate_interest(principal, rate=0.04):
    return principal * rate
````

Call it without providing a rate:


````python
print(calculate_interest(1000))
````

Provide a different rate:


````python
print(calculate_interest(1000, 0.06))
````

The default is used only when the argument is omitted.

**For the function definitions in this lesson, put required parameters before
parameters with default values.**

In this example, `principal` is required, and `rate` has a default.
